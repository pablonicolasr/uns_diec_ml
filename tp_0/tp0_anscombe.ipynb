# Import libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Create a function
def plot_anscombe(Xn_list, Yn_list, title="Dataset I Anscombe – Recta de regresión cortando el eje Y"):

    X_POINTS = [16.0, 8.0, 16.0, 16.0]
    Y_POINTS = [8.0, 6.0, 10.0, 6.0]

    x_line = np.linspace(0, 20, 100)
    y_line = 3 + 0.5 * x_line
    plt.figure()
    plt.scatter(Xn_list, Yn_list)
    plt.scatter(X_POINTS, Y_POINTS, color="red")
    plt.plot(x_line, y_line)
    
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(title)
    
    # Ejes desde 0
    plt.xlim(0, 20)
    plt.ylim(0, 20)

    plt.xticks(np.arange(0, 21, 1))
    plt.yticks(np.arange(0, 21, 1))
    
    plt.grid(True)
    plt.show()
```

```python
# Represent dataset I
x1_data = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y1_data = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
plot_anscombe(x1_data, y1_data, "Dataset I Anscombe – Recta de regresión cortando el eje Y")
```

```python
# Represent dataset II
x2_data = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y2_data = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
plot_anscombe(x2_data, y2_data, "Dataset II Anscombe – Recta de regresión cortando el eje Y")
```

```python
# Represent dataset III
x3_data = [10.0, 8.0, 13.0, 9.0, 11.0, 14.0, 6.0, 4.0, 12.0, 7.0, 5.0]
y3_data = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 6.73]
plot_anscombe(x3_data, y3_data, "Dataset I Anscombe – Recta de regresión cortando el eje Y")
```

```python
# Represent dataset IV
x4_data = [8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 8.0, 19.0, 8.0, 8.0, 8.0]
y4_data = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.5, 5.56, 7.91, 6.89]
plot_anscombe(x4_data, y4_data, "Dataset IV Anscombe – Recta de regresión cortando el eje Y")
```
